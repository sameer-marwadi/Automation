###Script that goes through all files in a specified directory, creates subdirectories based on file extensions, and moves each file to its corresponding subdirectory

```python
from pathlib import Path
import shutil

# Specify the directory to organize
DIRECTORY = r"C:/Users/12146/samtest"  # Change this to your folder

def organize_by_extension(directory):
    directory = Path(directory)

    if not directory.exists() or not directory.is_dir():
        print(f"Error: '{directory}' is not a valid directory.")
        return

    for item in directory.iterdir():
        # Skip directories
        if item.is_dir():
            continue

        # Determine the extension
        extension = item.suffix.lower().lstrip(".")

        # Files with no extension go into a "no_extension" folder
        if not extension:
            extension = "no_extension"

        # Create the destination folder if it doesn't exist
        destination_folder = directory / extension
        destination_folder.mkdir(exist_ok=True)

        # Move the file
        destination = destination_folder / item.name

        # Handle duplicate filenames
        counter = 1
        while destination.exists():
            destination = destination_folder / f"{item.stem}_{counter}{item.suffix}"
            counter += 1

        shutil.move(str(item), str(destination))
        print(f"Moved: {item.name} -> {destination_folder.name}/")

    print("Done!")

if __name__ == "__main__":
    organize_by_extension(DIRECTORY)
```
